```python
# ==========================================================
# 🎨 Donut Charts for Key Audio Features
# ==========================================================

# Function to create a donut chart
def donut_chart(counts, title):
    plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90, pctdistance=0.85)
    centre_circle = plt.Circle((0,0),0.70,fc='white')
    plt.gca().add_artist(centre_circle)
    plt.title(title)
    plt.show()

# -------------------------
# 1️⃣ Danceability
df['dance_level'] = pd.cut(df['danceability'], bins=[0, 0.4, 0.7, 1.0], labels=['Low', 'Medium', 'High'])
donut_chart(df['dance_level'].value_counts(), "Danceability Levels Distribution")

# -------------------------
# 2️⃣ Energy
df['energy_level'] = pd.cut(df['energy'], bins=[0, 0.4, 0.7, 1.0], labels=['Low', 'Medium', 'High'])
donut_chart(df['energy_level'].value_counts(), "Energy Levels Distribution")

# -------------------------
# 3️⃣ Valence
df['valence_level'] = pd.cut(df['valence'], bins=[0, 0.4, 0.7, 1.0], labels=['Low', 'Medium', 'High'])
donut_chart(df['valence_level'].value_counts(), "Valence Levels Distribution")

# -------------------------
# 4️⃣ Tempo (BPM)
df['tempo_level'] = pd.cut(df['tempo'], bins=[0, 90, 120, df['tempo'].max()], labels=['Slow', 'Medium', 'Fast'])
donut_chart(df['tempo_level'].value_counts(), "Tempo Levels Distribution")

# -------------------------
# 5️⃣ Popularity
df['popularity_level'] = pd.cut(df['popularity'], bins=[0, 40, 70, 100], labels=['Low', 'Medium', 'High'])
donut_chart(df['popularity_level'].value_counts(), "Popularity Levels Distribution")

# -------------------------
# 6️⃣ Acousticness
df['acoustic_level'] = pd.cut(df['acousticness'], bins=[0, 0.4, 0.7, 1.0], labels=['Low', 'Medium', 'High'])
donut_chart(df['acoustic_level'].value_counts(), "Acousticness Levels Distribution")

```

```python
# Calculate the popularity quartiles
pop_q = df['popularity'].quantile([0.25, 0.5, 0.75])
high_pop = df[df['popularity'] >= pop_q[0.75]]  # top 25% popular songs
print(f"Popularity threshold for top quartile: {pop_q[0.75]}")
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='danceability', 
    y='energy', 
    hue='valence', 
    size='popularity',
    data=high_pop, 
    palette='coolwarm',
    alpha=0.8
)
plt.title("Danceability vs Energy vs Valence for Highly Popular Songs")
plt.xlabel("Danceability")
plt.ylabel("Energy")
plt.legend(title='Valence', bbox_to_anchor=(1.05,1))
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='acousticness', 
    y='instrumentalness', 
    hue='speechiness', 
    size='popularity', 
    data=high_pop, 
    palette='viridis', 
    alpha=0.7
)
plt.title("Acousticness vs Instrumentalness vs Speechiness for Top Songs")
plt.xlabel("Acousticness")
plt.ylabel("Instrumentalness")
plt.legend(title='Speechiness', bbox_to_anchor=(1.05,1))
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='tempo', 
    y='loudness', 
    hue='mode',  # 0=Minor, 1=Major
    size='popularity', 
    data=high_pop, 
    palette='Set1',
    alpha=0.7
)
plt.title("Tempo vs Loudness vs Mode for Highly Popular Songs")
plt.xlabel("Tempo (BPM)")
plt.ylabel("Loudness (dB)")
plt.legend(title='Mode', bbox_to_anchor=(1.05,1))
plt.show()
```

```python
top_langs = high_pop['language'].value_counts().head(5).index
lang_avg = high_pop[high_pop['language'].isin(top_langs)].groupby('language')[['danceability','energy','valence']].mean()
lang_avg.plot(kind='bar', figsize=(12,6))
plt.title("Average Danceability, Energy, Valence by Language for Popular Songs")
plt.ylabel("Average Value")
plt.xlabel("Language")
plt.show()
```

```python
key_mode_time_avg = high_pop.groupby(['key','mode','time_signature'])['popularity'].mean().reset_index()
plt.figure(figsize=(12,6))
sns.scatterplot(
    x='key', 
    y='popularity', 
    hue='mode', 
    size='time_signature',
    data=key_mode_time_avg, 
    palette='Set2',
    alpha=0.8
)
plt.title("Popularity by Key, Mode, and Time Signature")
plt.xlabel("Key (0=C, 1=C#, ... 11=B)")
plt.ylabel("Average Popularity")
plt.legend(title='Mode / Time Signature', bbox_to_anchor=(1.05,1))
plt.show()
```

```python
# Create decade column
df['decade'] = (df['year']//10)*10
# Make sure the column exists in high_pop
high_pop['decade'] = (high_pop['year']//10)*10

# Then group by decade
# high_pop_decade = high_pop.groupby('decade')[['duration_ms','liveness']].mean().reset_index()

high_pop_decade = high_pop.groupby('decade')[['duration_ms','liveness']].mean().reset_index()

fig, ax1 = plt.subplots(figsize=(12,5))

color = 'tab:blue'
ax1.set_xlabel('Decade')
ax1.set_ylabel('Average Duration (ms)', color=color)
ax1.plot(high_pop_decade['decade'], high_pop_decade['duration_ms'], color=color, marker='o', label='Duration')
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Average Liveness', color=color)
ax2.plot(high_pop_decade['decade'], high_pop_decade['liveness'], color=color, marker='s', label='Liveness')
ax2.tick_params(axis='y', labelcolor=color)

plt.title("Average Duration and Liveness of Highly Popular Songs by Decade")
fig.tight_layout()
plt.show()
```
